# 数独の盤面生成

## 関数定義
ランダム生成を行う

```python
import random
import numpy as np

def is_valid(board, row, col, num):
    for i in range(9):
        if board[row][i] == num or board[i][col] == num:
            return False
    start_row, start_col = 3*(row//3), 3*(col//3)
    for i in range(3):
        for j in range(3):
            if board[start_row+i][start_col+j] == num:
                return False
    return True

def solve(board):
    for row in range(9):
        for col in range(9):
            if board[row][col] == 0:
                nums = list(range(1, 10))
                random.shuffle(nums)
                for num in nums:
                    if is_valid(board, row, col, num):
                        board[row][col] = num
                        if solve(board):
                            return True
                        board[row][col] = 0
                return False
    return True

def generate_full_board():
    board = [[0]*9 for _ in range(9)]
    solve(board)
    return board
```

いくつかを削除して問題にする

```python
def make_puzzle(board, remove_count=40):
    puzzle = [row.copy() for row in board]
    attempts = 0
    while attempts < remove_count:
        row, col = random.randint(0, 8), random.randint(0, 8)
        if puzzle[row][col] != 0:
            puzzle[row][col] = 0
            attempts += 1
    return puzzle
```

## OpenCVで描画

```python
import cv2

def draw_sudoku_image(board, filename="sudoku.png", cell_size=60):
    img_size = cell_size * 9
    img = np.ones((img_size, img_size), dtype=np.uint8) * 255

    for i in range(9):
        for j in range(9):
            val = board[i][j]
            if val != 0:
                text = str(val)
                x = j * cell_size + cell_size // 4
                y = i * cell_size + int(cell_size * 0.75)
                cv2.putText(img, text, (x, y), cv2.FONT_HERSHEY_SIMPLEX,
                            1, (0,), 2, cv2.LINE_AA)

    # グリッド線を描画
    for i in range(10):
        thickness = 2 if i % 3 == 0 else 1
        cv2.line(img, (0, i * cell_size), (img_size, i * cell_size), (0,), thickness)
        cv2.line(img, (i * cell_size, 0), (i * cell_size, img_size), (0,), thickness)

    cv2.imwrite(filename, img)
    return img
```

```python
full_board = generate_full_board()
puzzle = make_puzzle(full_board, remove_count=40)
draw_sudoku_image(puzzle, "sudoku_sample.png")
```

## 余白付きにする

```python
# 必要なモジュールの再インポート（エラー対処）
import numpy as np
import cv2
import random

# 余白付きで数独の盤面画像を描画する関数
def draw_sudoku_image_with_margin(board, filename="sudoku_margin.png", cell_size=60, margin=30):
    grid_size = cell_size * 9
    img_size = grid_size + margin * 2
    img = np.ones((img_size, img_size), dtype=np.uint8) * 255

    # 数字を描画
    for i in range(9):
        for j in range(9):
            val = board[i][j]
            if val != 0:
                text = str(val)
                x = margin + j * cell_size + cell_size // 4
                y = margin + i * cell_size + int(cell_size * 0.75)
                cv2.putText(img, text, (x, y), cv2.FONT_HERSHEY_SIMPLEX,
                            1, (0,), 2, cv2.LINE_AA)

    # グリッド線を描画
    for i in range(10):
        thickness = 2 if i % 3 == 0 else 1
        offset = margin
        cv2.line(img, (offset, offset + i * cell_size), (offset + grid_size, offset + i * cell_size), (0,), thickness)
        cv2.line(img, (offset + i * cell_size, offset), (offset + i * cell_size, offset + grid_size), (0,), thickness)

    cv2.imwrite(filename, img)
    return img

# 数独盤面生成関数
def is_valid(board, row, col, num):
    for i in range(9):
        if board[row][i] == num or board[i][col] == num:
            return False
    start_row, start_col = 3*(row//3), 3*(col//3)
    for i in range(3):
        for j in range(3):
            if board[start_row+i][start_col+j] == num:
                return False
    return True

def solve(board):
    for row in range(9):
        for col in range(9):
            if board[row][col] == 0:
                nums = list(range(1, 10))
                random.shuffle(nums)
                for num in nums:
                    if is_valid(board, row, col, num):
                        board[row][col] = num
                        if solve(board):
                            return True
                        board[row][col] = 0
                return False
    return True

def generate_full_board():
    board = [[0]*9 for _ in range(9)]
    solve(board)
    return board

# ランダムな数独盤面を生成して余白付き画像として出力
board = generate_full_board()
board = make_puzzle(board)
image_with_margin = draw_sudoku_image_with_margin(board)
```
